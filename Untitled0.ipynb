```python
import pandas as pd

# 1. Carregar os dados diretamente do link web (sem precisar baixar arquivo!)
url = 'https://raw.githubusercontent.com/IBM/telco-customer-churn-on-icp4d/master/data/Telco-Customer-Churn.csv'
df = pd.read_csv(url)

# 2. Limpeza simples dos dados
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
df['TotalCharges'].fillna(df['TotalCharges'].median(), inplace=True)

# 3. Análise dos dados (Perguntas de Negócio)
print("=== 1. TAXA GERAL DE CHURN (%) ===")
taxa_churn = df['Churn'].value_counts(normalize=True) * 100
print(taxa_churn.round(2))

print("\n=== 2. CHURN POR TIPO DE CONTRATO (%) ===")
churn_contrato = pd.crosstab(df['Contract'], df['Churn'], normalize='index') * 100
print(churn_contrato.round(2))

print("\n=== 3. GASTO MÉDIO MENSAL ($) ===")
gasto_medio = df.groupby('Churn')['MonthlyCharges'].mean()
print(gasto_medio.round(2))
```
